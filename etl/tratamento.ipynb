```python
df.rename(columns= {
    "id": "product_id",
    "title": "product_name",
    "description": "description",
    "category": "category",
    "price": "price",
    "discountPercentage": "discount_percentage",
    "rating": "rating",
    "stock": "stock",
    "tags": "tags",
    "brand": "brand",
    "sku": "sku",
    "weight": "weight",
    "availabilityStatus": "availability_status",
    "minimumOrderQuantity": "minimum_order_quantity",
    "dimensions.width": "width",
    "dimensions.height": "height",
    "dimensions.depth": "depth",
    "meta.createdAt": "created_at",
    "meta.updatedAt": "updated_at",
    "meta.barcode": "barcode",
    "meta.qrCode": "qr_code"
}, inplace=True)
```

```python
df.isnull().sum()
```

```python
df['brand'] = df['brand'].fillna("Unknown")
```

```python
df.drop_duplicates
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns=["barcode", "qr_code", "width", "height", "depth"], inplace=True)
```

```python
df.head()
```

```python
df["tags"] = df["tags"].apply(lambda x: ", ".join(x))
```

```python
df.info()
```

```python
df['minimum_order_quantity'] = df['minimum_order_quantity'].astype(int)
df['created_at'] = pd.to_datetime(df['created_at'])
df['updated_at'] = pd.to_datetime(df['updated_at'])
df["price"] = df["price"].astype(float).round(2)
df["discount_percentage"] = df["discount_percentage"].astype(float).round(2)
df["rating"] = df["rating"].astype(float).round(2)
df["weight"] = df["weight"].astype(float)
```

```python
df.info()
```

```python
from datetime import datetime, timezone

df["collection_date"] = datetime.now(timezone.utc)
```

```python
#SEPARANDO EM DATAFRAMES PARA ADICIONAR NO BANCO(FORMA MAIS ORGANIZADA E LIMPA)
```

```python
df_products = df[
    [
        "product_id",
        "product_name",
        "description",
        "category",
        "brand",
        "sku",
        "weight",
        "minimum_order_quantity",
        "tags",
        "created_at"
    ]
]
```

```python
df_price_history = df[
    [
        "product_id",
        "price",
        "discount_percentage",
        "updated_at",
        "collection_date"
    ]
]
```

```python
df_stock_history = df[
    [
        "product_id",
        "stock",
        "availability_status",
        "collection_date"
    ]
]
```

```python
#ADICIONANDO DADOS NA TABELA
```

```python
df_products.to_sql(
    name="products",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
df_price_history.to_sql(
    name = "price_history",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
df_stock_history.to_sql(
    name = "stock_history",
    con=engine,
    if_exists="append",
    index=False
)
```
